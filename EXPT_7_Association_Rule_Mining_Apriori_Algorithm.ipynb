```python
from itertools import combinations

def get_support(itemset, transactions):
    """Calculate support of an itemset"""
    count = sum(1 for t in transactions if itemset.issubset(t))
    return count, count / len(transactions)

def apriori(transactions, min_support, min_confidence):
    # Step 1: Generate 1-itemsets
    items = set().union(*transactions)
    itemsets = [{i} for i in items]

    # Store frequent itemsets
    freq_itemsets = []

    # Iteratively generate k-itemsets
    k = 1
    while itemsets:
        new_freq = []
        for itemset in itemsets:
            count, support = get_support(itemset, transactions)
            if support >= min_support:
                new_freq.append((itemset, support))
        freq_itemsets.extend(new_freq)

        # Generate next level itemsets
        k += 1
        itemsets = [set(a) | set(b) for a in items for b in items
                    if len(set(a) | set(b)) == k]

        # Remove duplicates
        itemsets = list(map(set, set(frozenset(i) for i in itemsets)))

    # Step 2: Generate rules
    rules = []
    for itemset, support in freq_itemsets:
        if len(itemset) > 1:
            for i in range(1, len(itemset)):
                for lhs in combinations(itemset, i):
                    lhs, rhs = set(lhs), itemset - set(lhs)
                    lhs_count, lhs_support = get_support(lhs, transactions)
                    if lhs_support > 0:
                        confidence = support / lhs_support
                        if confidence >= min_confidence:
                            rules.append((lhs, rhs, support, confidence))
    return freq_itemsets, rules


if __name__ == "__main__":
    # Example: take dataset from user input
    print("Enter transactions (comma separated items per line, blank line to end):")
    transactions = []
    while True:
        line = input().strip()
        if not line:
            break
        transactions.append(set(line.split(",")))

    min_support = float(input("Enter minimum support (0-1): "))
    min_confidence = float(input("Enter minimum confidence (0-1): "))

    freq_itemsets, rules = apriori(transactions, min_support, min_confidence)

    print("\nFrequent Itemsets:")
    for itemset, support in freq_itemsets:
        print(f"{itemset} -> support: {support:.2f}")

    print("\nAssociation Rules:")
    for lhs, rhs, support, confidence in rules:
        print(f"{lhs} -> {rhs} | support={support:.2f}, confidence={confidence:.2f}")
```
